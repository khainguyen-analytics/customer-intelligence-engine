## 2. Load the dataset

The Online Retail II dataset contains transactions from December 2009 to 2011.

The Excel file contains two worksheets:
- Year 2009-2010
- Year 2010-2011

We will load both worksheets and combine them into one DataFrame.

```python
# Define the dataset location
file_path = Path("/Users/richardnguyenk/Downloads/GitHub Project/customer-intelligence-engine/data/raw/online_retail_II.xlsx")

# Check whether the file exists
if not file_path.exists():
    raise FileNotFoundError(f"Dataset not found: {file_path}")

# Read both worksheets
retail_data = pd.read_excel(
    file_path,
    sheet_name=None,
    engine="openpyxl"
)

# Combine both years
df = pd.concat(
    retail_data.values(),
    ignore_index=True
)

print("Dataset loaded successfully.")
```

## 7. Investigating Data Quality Issues

```python
# Inspect missing Customer IDS

df[df["Customer ID"].isnull()].head()
```

```python
# Checking for how many negative quantities belong to cancelled invoices

# Select transactions with negative quanntities
negative_sales = df[df["Quantity"] < 0]

# Count how many have cancelled invoice numbers
negative_sales["Invoice"].astype(str).str.startswith("C").sum()
```

```python
# How many missing Customer IDs belong to postive-quantity sales?

# Select records with missing Customer IDs
missing_customers = df[df["Customer ID"].isnull()]

# Count how many have positive quantities and prices
missing_customers[(missing_customers["Quantity"] > 0) & (missing_customers["Price"] > 0)].shape[0]
```

## 8. Initial Data Quality Findings

The initial data exploration identified several data quality issues in the Online Retail II dataset.

There are 34,335 duplicate rows and 243,007 records with missing Customer IDs. However, 236,122 records with missing Customer IDs have positive quantities and prices, suggesting that many may still represent valid sales.

We also identified 22,950 records with negative quantities. Of these, 19,493 are associated with cancelled invoices, while the remaining 3,457 require further investigation.

Based on these findings, the cleaning process will separate completed sales from cancellations and other unusual transactions. Records without Customer IDs will be retained for overall sales analysis but excluded from customer-level analysis.

The original dataset will remain unchanged.
